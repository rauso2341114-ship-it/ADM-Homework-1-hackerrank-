```python
'''String Formatting'''
def print_formatted(number):
   
    
    width = len(bin(number)[2:])
    
    for i in range(1, number + 1):
        dec = str(i).rjust(width)
        octal = oct(i)[2:].rjust(width)
        hexa = hex(i)[2:].upper().rjust(width)
        binary = bin(i)[2:].rjust(width)
        
        print(f"{dec} {octal} {hexa} {binary}")
        
if __name__ == '__main__':
    n = int(input())
    print_formatted(n)
```

```python
'''Alphabet rangoli'''

def print_rangoli(size):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    design = []
    
    for i in range(size):
        s = "-".join(alphabet[i:size])
        row = (s[::-1] + s[1:]).center(4 * size - 3, "-")
        design.append(row)
    print("\n".join(design[::-1] + design[1:]))
   

if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
```

```python
'''The Minion Game'''

def minion_game(string):
    vowels = "AEIOU"
    kevin_score = 0
    stuart_score = 0
    n = len(string)
    
    for i in range(n):
        if string[i] in vowels:
            kevin_score += n - i
        else:
            stuart_score += n - i
            
    if kevin_score > stuart_score:
        print(f"Kevin {kevin_score}")
    elif kevin_score==stuart_score:
        print("Draw")
    else:
        print(f"Stuart {stuart_score}")

if __name__ == '__main__':
    s = input()
    minion_game(s)
```

```python
'''Merge the tools!'''

def merge_the_tools(string, k):
   for i in range(0, len(string), k):
        substring = string[i:i + k]
        saw = set()
        results = []
        for charact in substring:
            if charact not in saw:
                saw.add(charact)
                results.append(charact)
        print("".join(results))

if __name__ == '__main__':
    string, k = input(), int(input())
    merge_the_tools(string, k)
```

```python
'''Collections.deque()'''
from collections import deque

n = int(input())
dq = deque()

for _ in range(n):
    cmd = input().split()
    op = cmd[0]
    
    if op == "append":
        dq.append(int(cmd[1]))
    elif op == "appendleft":
        dq.appendleft(int(cmd[1]))
    elif op == "pop":
        dq.pop()
    elif op == "popleft":
        dq.popleft()

print(*(dq))
```

```python
'''Piling Up!'''

from collections import deque

for _ in range(int(input())):
    input()
    dq = deque(map(int, input().split()))
    lim = float('inf')
    ok = True
    while dq:
        val = dq.popleft() if dq[0] >= dq[-1] else dq.pop()
        if val <= lim:
            lim = val
        else:
            ok = False
            break
    print("Yes" if ok else "No")
```

```python
'''Time Delta'''

#!/bin/python3

import math
import os
import random
import re
import sys
from datetime import datetime

# Complete the time_delta function below.
def time_delta(t1, t2):
    fmt = "%a %d %b %Y %H:%M:%S %z"
    d1 = datetime.strptime(t1, fmt)
    d2 = datetime.strptime(t2, fmt)
    return str(int(abs((d1 - d2).total_seconds())))
    
if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    t = int(input())

    for t_itr in range(t):
        t1 = input()

        t2 = input()

        delta = time_delta(t1, t2)

        fptr.write(delta + '\n')

    fptr.close()


```

```python
'''HTML Parser - Part 1'''
from html.parser import HTMLParser

class P(HTMLParser):

    def handle_starttag(self, t, a):
        print("Start :", t)
        for k, v in a:
            print(f"-> {k} > {v}")

    def handle_endtag(self, t):
        print("End   :", t)

    def handle_startendtag(self, t, a):
        print("Empty :", t)
        for k, v in a:
            print(f"-> {k} > {v}")


p = P()
h = "".join([input() for _ in range(int(input()))])
p.feed(h)
```

```python
'''HTML Parser - Part 2'''

from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_comment(self, d):
        if "\n" in d:
            print(">>> Multi-line Comment")
        else:
            print(">>> Single-line Comment")
        print(d)

    def handle_data(self, d):
        if d != "\n":
            print(">>> Data")
            print(d)
  
html = ""       
for i in range(int(input())):
    html += input().rstrip()
    html += '\n'
    
parser = MyHTMLParser()
parser.feed(html)
parser.close()
```

```python
'''Validating UID'''

import re

for _ in range(int(input())):
    s = input()
    if (
        len(s) == 10
        and len(set(s)) == 10
        and bool(re.match(r"^[a-zA-Z0-9]*$", s))
        and len(re.findall(r"[A-Z]", s)) >= 2
        and len(re.findall(r"\d", s)) >= 3
    ):
        print("Valid")
    else:
        print("Invalid")
```

```python
'''Validating Credit Card Numbers'''

for _ in range(int(input())):
    s = input()
    ok = True
    if "-" in s:
        p = s.split("-")
        if len(p) != 4 or any(len(x) != 4 for x in p):
            ok = False
        c = "".join(p)
    else:
        c = s
    
    if ok and (len(c) != 16 or not c.isdigit() or c[0] not in "456"):
        ok = False
        
    if ok:
        for i in range(13):
            if c[i] == c[i+1] == c[i+2] == c[i+3]:
                ok = False
                break
                
    print("Valid" if ok else "Invalid")
```

```python
''' XML 1 - find the score'''

import sys
import xml.etree.ElementTree as etree

def get_attr_number(node):
  return sum(len(x.attrib) for x in node.iter())

if __name__ == '__main__':
    sys.stdin.readline()
    xml = sys.stdin.read()
    tree = etree.ElementTree(etree.fromstring(xml))
    root = tree.getroot()
    print(get_attr_number(root))
```

```python
'''XML2 - Find the Maximum Depth'''

import xml.etree.ElementTree as etree

maxdepth = 0
def depth(elem, level):
    global maxdepth
    
    level += 1
    if level > maxdepth:
        maxdepth = level
    for ch in elem:
        depth(ch, level)
        
        
if __name__ == '__main__':
    n = int(input())
    xml = ""
    for i in range(n):
        xml =  xml + input() + "\n"
    tree = etree.ElementTree(etree.fromstring(xml))
    depth(tree.getroot(), -1)
    print(maxdepth)
```

```python
'''Standardize Mobile Number Using Decorators'''

def wrapper(f):
    def fun(l):
        f([f"+91 {x[-10:-5]} {x[-5:]}" for x in l])
    return fun

@wrapper
def sort_phone(l):
    print(*sorted(l), sep='\n')

if __name__ == '__main__':
    l = [input() for _ in range(int(input()))]
    sort_phone(l) 


```

```python
''' Decorators 2 - Name Directory '''

import operator

def person_lister(f):
    def inner(people):
        return [f(p) for p in sorted(people, key=lambda x: int(x[2]))]
    return inner

@person_lister
def name_format(person):
    return ("Mr. " if person[3] == "M" else "Ms. ") + person[0] + " " + person[1]

if __name__ == '__main__':
    people = [input().split() for i in range(int(input()))]
    print(*name_format(people), sep='\n')

```

```python
''' kangaroo --> Number Line Jumps '''


#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'kangaroo' function below.
#
# The function is expected to return a STRING.
# The function accepts following parameters:
#  1. INTEGER x1
#  2. INTEGER v1
#  3. INTEGER x2
#  4. INTEGER v2
#

def kangaroo(x1, v1, x2, v2):
  
    if v1 <= v2:
        return "NO"

    while x1 < x2:
        x1 += v1
        x2 += v2
        
        if x1 == x2:
            return "YES"

    return "NO"
    
    
    
if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    first_multiple_input = input().rstrip().split()

    x1 = int(first_multiple_input[0])

    v1 = int(first_multiple_input[1])

    x2 = int(first_multiple_input[2])

    v2 = int(first_multiple_input[3])

    result = kangaroo(x1, v1, x2, v2)

    fptr.write(result + '\n')

    fptr.close()
```

```python
'''Strange advertising ---> Viral advertising'''

import math
import os
import random
import re
import sys

#
# Complete the 'viralAdvertising' function below.
#
# The function is expected to return an INTEGER.
# The function accepts INTEGER n as parameter.
#

def viralAdvertising(n):
    shared = 5
    cumulative_likes = 0
    
    for _ in range(n):
        liked = shared // 2
        cumulative_likes += liked
        shared = liked * 3
        
    return cumulative_likes

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    n = int(input().strip())

    result = viralAdvertising(n)

    fptr.write(str(result) + '\n')

    fptr.close()
```

```python
'''Recursive Digit Sum '''

#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'superDigit' function below.
#
# The function is expected to return an INTEGER.
# The function accepts following parameters:
#  1. STRING n
#  2. INTEGER k
#

def superDigit(n, k):

    if len(n) == 1 and k == 1:
        return int(n)
    
    total = sum(int(d) for d in n) * k
    return superDigit(str(total), 1)

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    first_multiple_input = input().rstrip().split()

    n = first_multiple_input[0]

    k = int(first_multiple_input[1])

    result = superDigit(n, k)

    fptr.write(str(result) + '\n')

    fptr.close()
```

```python
'''Insertion Sort - Part 2'''

#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'insertionSort2' function below.
#
# The function accepts following parameters:
#  1. INTEGER n
#  2. INTEGER_ARRAY arr
#


def insertionSort2(n, arr):
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
            
        arr[j + 1] = key
        print(*arr)

if __name__ == '__main__':
    n = int(input().strip())

    arr = list(map(int, input().rstrip().split()))

    insertionSort2(n, arr)
```
